# Soul Cradle SDK — live demo

Send a document, get back a **hash-verified, multi-agent evaluation**: the eight assessor-witnesses hear the action *"rely on this document,"* judge it on observed evidence, and every judgment is sealed.

The honest contract: the payload proves the evaluation is **unaltered**, not that the document is **true**.

```
!git clone -q https://github.com/herbievelezjr/Soul_Cradle.git
!pip install -q pypdf
import sys
sys.path += ['/content/Soul_Cradle', '/content/Soul_Cradle/sdk']
from soul_cradle_sdk import evaluate_document, verify_payload
print('SDK ready')
```

## 1. The risky memo — "guaranteed 100% risk-free returns", no methodology

```
risky = evaluate_document('/content/Soul_Cradle/sdk/examples/risky_memo.md')
print('Overall verdict:', risky['overall_verdict'])
for s in risky['sections']:
    flagged = [j['assessor_id'] for j in s['judgments'] if j['verdict'] == 'flagged']
    print(f"  [{s['section_id']}] {s['panel_verdict']:9s} {s['heading'][:50]}")
    if flagged:
        print(f"             flagged by: {', '.join(flagged)}")
```

## 2. The clean policy — methodology, citations, limitations stated

```
clean = evaluate_document('/content/Soul_Cradle/sdk/examples/clean_policy.md')
print('Overall verdict:', clean['overall_verdict'])
for s in clean['sections']:
    print(f"  [{s['section_id']}] {s['panel_verdict']:9s} {s['heading'][:50]}")
```

## 3. Verify the seal — and inspect the dissent

A contested verdict means witnesses disagreed. The disagreement is the signal: here is exactly what the flagging witnesses saw.

```
import json
ok, details = verify_payload(risky)
print('Payload verified:', ok)
sec = risky['sections'][0]
print(json.dumps(sec['dissent'], indent=2)[:1500])
```

## 4. Try your own PDF

Upload any text-based PDF (scanned/image PDFs have no extractable text and are rejected rather than guessed at).

```
from google.colab import files
up = files.upload()
for name in up:
    p = evaluate_document(f'/content/{name}')
    print(f'{name}: {p["overall_verdict"]}  (payload {p["payload_hash"][:16]}...)')
    ok, _ = verify_payload(p)
    print('verified:', ok)
```
